```python
def decode_glass_catalog(glass_cat_name, glass_lib):
    """Decode all of the glass names in glass_cat_name.

    Print out the original glass names and the decoded version side by side.
    Args:
        glass_cat_name (str): name of a supported glass catalog
        
    Returns:
        group_num (dict): all the non-unique group-num pairs
        prefixes (dict): all the non-null prefixes used
        suffixes (dict): all the non-null suffixes used
        """
    from opticalglass.util import Counter
    prefixes = Counter()
    suffixes = Counter()
    group_nums = Counter()
    glass_cat = gfact.og_glass_libs[glass_lib][glass_cat_name]
    glasses = glass_cat.get_glass_names()
    glass_list = []
    for gn in glasses:
        gla = decode_glass_name(gn)
        if gla.prefix != '':
            prefixes[gla.prefix] += 1
        if gla.suffix != '':
            suffixes[gla.suffix] += 1
        group_nums[gla.group_num] += 1
        glass_list.append((gla, gn, glass_cat_name, glass_lib))
        #print("{:14s} {:>2s}  {:8s}  {:12s}".format(gn, prefix, group_num, suffix))
    group_nums = {k:v for k,v in group_nums.items() if v>1}
    return group_nums, prefixes, suffixes, glass_list
```

```python
def int_cast(i: str) -> int:
    try:
        i_val = int(i)
    except ValueError as ve:
        i_val = int(float(i))
    return i_val
```

```python
def int_trunc(i: str) -> int:
    int_str = i if (idx:=i.find('.'))<0 else i[:idx]
    i_val = int(int_str) if len(int_str)>0 else 0
    return i_val
```
